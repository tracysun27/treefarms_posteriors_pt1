```python
# saving all the trees
def save_rset(model, config):
    NUM_TREES = model.get_tree_count()

    model_dict = {}
    for i in range(NUM_TREES):
        model_dict[i] = json.loads(model[i].json())

    with open(f'{path}/treeFarms/ttt_tf_rset_{config["regularization"]}.json', 'w') as file:
        json.dump(model_dict, file, indent=4)
```

```python
# code from varun to count number of leaves in tree

# Step 1: for the i^th tree in treefarms, 
# convert the tree from dict to a gosdt Node/Leaf object
from gosdt._tree import Node, Leaf

def dict_to_tree(tree_dict, X, y):
        # Base case: if the dictionary represents a prediction, return a Leaf object
        if "prediction" in tree_dict:
            return Leaf(prediction=tree_dict["prediction"], loss=(y != int(tree_dict["prediction"])).sum())#/self.n
        # Recursive case: construct a Node object
        # Recursively build the true and false branches
        left_child = dict_to_tree(
            tree_dict["true"], X[X.iloc[:, tree_dict["feature"]] == 1], y[X.iloc[:, tree_dict["feature"]] == 1])
        right_child = dict_to_tree(
            tree_dict["false"], X[X.iloc[:, tree_dict["feature"]] == 0], y[X.iloc[:, tree_dict["feature"]] == 0])
        node = Node(feature=tree_dict["feature"],
                    left_child=left_child, right_child=right_child)
        return node

# Step 2: Compute the # leaves in this object
def get_num_leaves(model):
    if isinstance(model, Leaf):
        return 1
    return get_num_leaves(model.left_child) + get_num_leaves(model.right_child)
```

```python
# changed training loss initializing fxn bc it was giving an error
def my_initialize_training_loss(tree, X, y):
    for node in tree.__all_leaves__():
        node["loss"] = 0.0
    (n, m) = X.shape
    for i in range(n):
        node = tree.__find_leaf__(X.values[i, :])
        label = y.values[i]
        weight = 1 / n
        if "loss" not in node:
            node["loss"] = 0.0
        if node["prediction"] != label:
            node["loss"] += weight
        # print(node)
        # print(node["loss"])
```

```python
def get_objs_tf(model, X, y):
    NUM_TREES = model.get_tree_count()

    num_leaves_per_tree = {}
    loss_per_tree = {}
    combined_obj_per_tree = {}

    for i in range(NUM_TREES):
        tree_dict = vars(model[i])['source']
        gosdt_model = dict_to_tree(tree_dict, X, y)
        num_leaves_per_tree[i] = get_num_leaves(gosdt_model)
        
        saved_tree= model.__getitem__(i)
        my_initialize_training_loss(saved_tree, X, y)

        loss_per_tree[i] = saved_tree.loss()
        combined_obj_per_tree[i] = loss_per_tree[i] + (config2['regularization']*num_leaves_per_tree[i]) 
    return combined_obj_per_tree
```

## MAPtree objectives

```python
with open(f"{path}/treeFarms/maptree_dict.json", 'r') as file:
    maptree_dict = json.load(file)

with open(f"{path}/treeFarms/maptree_dict_train.json", 'r') as file:
    maptree_dict_train = json.load(file)

with open(f"{path}/treeFarms/maptree_alpha_beta_res_dict_train.json", 'r') as file:
    maptree_ab_dict_train = json.load(file)

with open(f"{path}/treeFarms/maptree_rho_res_train_dict.json", 'r') as file:
    maptree_rho_dict_train = json.load(file)
```

```python
maptree_params_dict_train = maptree_ab_dict_train + maptree_rho_dict_train
```

```python
def relabel(tree_dict):
    if not isinstance(tree_dict, dict):
        return tree_dict

    new_dict = {}
    for key, value in tree_dict.items():
        if key == "left":
            new_dict["false"] = relabel(value)
        elif key == "right":
            new_dict["true"] = relabel(value)
        else:
            new_dict[key] = relabel(value)
    return new_dict
```

```python
def get_maptree_obj(config, maptree_dict, X, y):
        gosdt_maptree_model = dict_to_tree(maptree_dict, X, y)
        maptree_num_leaves = get_num_leaves(gosdt_maptree_model)
        maptree_tf_class = TreeClassifier(relabel(maptree_dict))
        my_initialize_training_loss(maptree_tf_class, X, y)

        maptree_saved_loss = maptree_tf_class.loss()
        maptree_tf_obj = maptree_saved_loss + (config['regularization']*maptree_num_leaves) 
        return maptree_tf_obj
```

```python
reg_list = [0.01, 0.012, 0.0075]

for my_maptree in maptree_params_dict_train:
    if not isinstance(my_maptree.get("tf_obj"), dict):
        my_maptree["tf_obj"] = {}
    print("Alpha:", my_maptree["alpha"])
    print("Beta:", my_maptree["beta"])
    print("Rho:", my_maptree["rho"])
    for reg in reg_list:
        my_config = {
        'regularization': reg,
        'rashomon_bound_multiplier': 0.05,
        'depth_budget': 5}
        tf_obj = get_maptree_obj(my_config, relabel(my_maptree["tree"]), ttt_X_train, ttt_y_train)
        print("Reg:", reg,  "Objective:", tf_obj)
        if "tf_obj" not in my_maptree:
            my_maptree["tf_obj"] = {str(reg): tf_obj}
        else:
            my_maptree["tf_obj"][str(reg)] = tf_obj
    print()
```
